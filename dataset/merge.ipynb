```python
import pandas as pd
import pyarrow.parquet as pq

# Load parquet
table = pq.read_table("train-00000-of-00001.parquet")
df1 = table.to_pandas()

# Load csv
df2 = pd.read_csv("chatbot_dataset.csv")

# Merge
merged_df = pd.concat(
    [df1, df2],
    ignore_index=True
)

# Save
merged_df.to_csv(
    "diabetes_merged_dataset.csv",
    index=False,
    encoding="utf-8"
)

print("Shape:", merged_df.shape)
```
